```python
from pathlib import Path

ROOT = Path(".").resolve()  # 在 odil-hackathon 仓库根目录运行 notebook
DATA = ROOT / "data"
IMG_DIR = DATA / "images"

DEV_TSV = DATA / "splits" / "dev.tsv"
CLASSES_TSV = DATA / "classes.tsv"

OUT_DIR = ROOT / "experiments" / "iconclass_style_infer" / "artifacts"
OUT_DIR.mkdir(parents=True, exist_ok=True)
DEV_PRED_JSONL = OUT_DIR / "dev_predictions.jsonl"

# Qwen3.0: 用 Qwen3-VL-* 才是图文模型
MODEL_NAME = "Qwen/Qwen3-VL-4B-Instruct"  # 或 "Qwen/Qwen3-VL-8B-Instruct"
```

```python
from transformers import pipeline
from PIL import Image
```

```python
pipe = pipeline(
    "image-text-to-text",
    model="Qwen/Qwen3-VL-2B-Instruct",
    trust_remote_code=True,
    device_map="auto"
)
```

```python
image = Image.open("data/images/9.jpg").convert("RGB")
```

```python
messages = [
    {
        "role": "user",
        "content": [
            {"type": "image", "image": image},
            {
                "type": "text",
                "text": (
                    'Extract TIMEL concepts for this image. '
                )
            }
        ]
    }
]
```

```python
output = pipe(
    messages,
    max_new_tokens=256,
    num_beams=4
)

print(output[0]["generated_text"])
```
